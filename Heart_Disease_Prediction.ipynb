importing the Dependencies

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

```

Data collection and Processing

```python
#loading  the csv data to a Pandas DataFrame
heart_data=pd.read_csv('/content/archive.zip')
```

1--->Defective Heart
0---->Healthy

Splitting the Features and Target

```python
# Replace 'Presence' and 'Absence' in 'Heart Disease' column with numerical values
# Fill any unmapped values (which become NaN) with 0 and convert to integer type
heart_data['Heart Disease'] = heart_data['Heart Disease'].map({'Presence': 1, 'Absence': 0}).fillna(0).astype(int)

X=heart_data.drop(columns='Thallium',axis=1)
Y=heart_data['Thallium']
```

```python
print(X)
```

```python
print(Y)
```

splitting the Data into Training data & Test Data


```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

Logistic Regression

```python
model=LogisticRegression()
```

```python
#training the LogisticRegression model with Training data

# Convert 'Heart Disease' column in X_train from string to numerical
X_train_processed = X_train.copy()
X_train_processed['Heart Disease'] = X_train_processed['Heart Disease'].map({'Presence': 1, 'Absence': 0}).fillna(0).astype(int)

model.fit(X_train_processed, Y_train)
```

Model Evaluation


Accuarcy Score

```python
#accuarcy on test data

# Convert 'Heart Disease' column in X_test from string to numerical for prediction
X_test_processed = X_test.copy()
X_test_processed['Heart Disease'] = X_test_processed['Heart Disease'].map({'Presence': 1, 'Absence': 0}).fillna(0).astype(int)

X_test_prediction=model.predict(X_test_processed)
test_data_accuarcy=accuracy_score(X_test_prediction,Y_test)
```

```python
print("Accuarcy onTest :",test_data_accuarcy)
```

Building a Predictive System

```python
input_data=(57,1,2,124,261,0,0,141,0,0.3,1,0,7)
#change the input data ta a numpy array
input_data_as_numpy_array=np.asarray(input_data)

#reshape the numpy array as we are predicting for only on instance
input_data_reshaped=input_data_as_numpy_array.reshape(1,-1)
prediction=model.predict(input_data_reshaped)
print(prediction)

if(prediction[0]==0):
  print("The person does not have Heart Disease")
else:
  print("The person has Heart Disease")
```
